```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
Q1 = df[["Quantity","UnitPrice","TotalPrice"]].quantile(0.25)
Q3 = df[["Quantity","UnitPrice","TotalPrice"]].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
df = df[
    ~(
        (
            (df[["Quantity","UnitPrice","TotalPrice"]] < lower)
            |
            (df[["Quantity","UnitPrice","TotalPrice"]] > upper)
        ).any(axis=1)
    )
]
```

```python
df.shape
```

```python
df["OrderMonth"] = df["Date"].dt.month
```

```python
df["OrderYear"] = df["Date"].dt.year
```

```python
df["AveragePricePerItem"] = df["TotalPrice"] / df["Quantity"]
```

```python
df.to_excel("Cleaned_Dataset.xlsx", index=False)
```
